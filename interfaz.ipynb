```python
import pandas as pd
import tensorflow as tf
import numpy as np
```

```python
archivo = input('Inserta el nombre del archivo: ')
extension = archivo.split('.')[1]
f = open(archivo, "r")
texto = f.read()

extensiones = ['cpp', 'cs', 'c']
lenguaje = extensiones.index(extension)
```

```python
data = {
    'language_name': [lenguaje],
    'code': [texto],
}
columnas = columns=data.keys()
datos = pd.DataFrame.from_dict(data)
datos.head()
```

```python
model = tf.keras.models.load_model('./checkpoints/c_cnn_best.h5')

y_pred_proba = model.predict(datos)
y_pred = np.argmax(y_pred_proba, axis=1)

class_names = ['Generado por IA', 'Hecho por humano']
y_pred_name = class_names[y_pred[0]]

print(y_pred_name)
```
